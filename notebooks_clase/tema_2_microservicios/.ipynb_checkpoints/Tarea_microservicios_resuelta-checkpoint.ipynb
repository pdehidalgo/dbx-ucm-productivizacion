# Ejercicio de cierre resuelto: microservicio de inferencia con serialización

Esta versión resuelta combina los conceptos de **FastAPI**, **Pydantic** y **serialización con `pickle`** sobre un ejemplo sencillo con `iris`.


## Objetivos

1. Entrenar un clasificador sobre el dataset `iris`.
2. Serializar el modelo entrenado en un fichero `.pkl`.
3. Deserializar el modelo para servirlo desde FastAPI.
4. Definir un esquema de entrada con Pydantic.
5. Exponer dos endpoints:
   - `GET /health`
   - `POST /predict`
6. Consumir la API desde un cliente Python con `requests`.


````python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import pickle
import threading

import nest_asyncio
import pandas as pd
import requests
import seaborn as sns
import uvicorn
from fastapi import FastAPI
from pydantic import BaseModel, field_validator
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

MODEL_DIR = Path("artifacts")
MODEL_DIR.mkdir(exist_ok=True)
MODEL_PATH = MODEL_DIR / "iris_api_model_v01.pkl"
FEATURE_NAMES = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
EXERCISE_PORT = 8001
````

## Apartado 1: entrenamiento del modelo


````python
iris = sns.load_dataset("iris")
iris.sample(5, random_state=99)

X = iris[FEATURE_NAMES]
y = iris["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=99,
)

model = DecisionTreeClassifier(random_state=99)
model.fit(X_train, y_train)

accuracy = model.score(X_test, y_test)
print(f"Accuracy en test: {accuracy:.4f}")
````

## Apartado 2: serialización y carga del modelo


````python
with open(MODEL_PATH, "wb") as file:
    pickle.dump(model, file)

with open(MODEL_PATH, "rb") as file:
    loaded_model = pickle.load(file)

example_df = pd.DataFrame([[6.9, 3.1, 5.4, 2.1]], columns=FEATURE_NAMES)
print("Predicción de prueba:", loaded_model.predict(example_df)[0])
````

## Apartado 3: contrato de entrada con Pydantic


````python
class IrisFeatures(BaseModel):
    sepal_length: float
    sepal_width: float
    petal_length: float
    petal_width: float

    @field_validator("sepal_length", "sepal_width", "petal_length", "petal_width")
    @classmethod
    def check_positive(cls, value):
        if value <= 0:
            raise ValueError("Todas las variables del iris deben ser positivas")
        return value
````

## Apartado 4: microservicio con FastAPI


````python
app = FastAPI(title="Iris Microservice Exercise")

with open(MODEL_PATH, "rb") as file:
    deployed_model = pickle.load(file)

@app.get("/health")
def healthcheck():
    return {
        "status": "ok",
        "model_loaded": deployed_model is not None,
        "model_path": str(MODEL_PATH),
    }

@app.post("/predict")
def predict(features: IrisFeatures):
    data = features.model_dump()
    X = pd.DataFrame([[data[name] for name in FEATURE_NAMES]], columns=FEATURE_NAMES)
    prediction = deployed_model.predict(X)[0]
    return {"prediction": prediction}
````

## Apartado 5: levantar el servidor

Si el puerto estuviera ocupado, cambia `EXERCISE_PORT` en la celda de imports.


````python
nest_asyncio.apply()

def run():
    uvicorn.run(app, host="127.0.0.1", port=EXERCISE_PORT)

thread = threading.Thread(target=run)
thread.start()

print(f"Servidor levantado en http://127.0.0.1:{EXERCISE_PORT}")
````

## Apartado 6: consumo de la API


````python
base_url = f"http://127.0.0.1:{EXERCISE_PORT}"

health_response = requests.get(f"{base_url}/health")
print("Health:", health_response.json())

payload = {
    "sepal_length": 6.9,
    "sepal_width": 3.1,
    "petal_length": 5.4,
    "petal_width": 2.1,
}

prediction_response = requests.post(f"{base_url}/predict", json=payload)
print("Predict:", prediction_response.json())
````

## Respuestas a las preguntas del notebook a completar

1. ¿Qué ventaja aporta cargar el modelo desde disco frente a reutilizar directamente la variable `model` entrenada en memoria? Cargar el modelo desde disco desacopla el entrenamiento del servicio. Permite reiniciar la API sin reentrenar, desplegar el modelo en otro proceso o máquina y validar que realmente servimos el artefacto persistido y no una variable de una sesion interactiva.
2. ¿Qué error devuelve FastAPI si falta un campo del JSON o si el tipo no coincide? FastAPI devuelve un `422 Unprocessable Entity` cuando el JSON no cumple el esquema de Pydantic, por ejemplo si falta un campo obligatorio o si un valor no puede convertirse al tipo esperado.
3. ¿Qué limitaciones tiene `pickle` para un despliegue serio en producción? no es seguro frente a ficheros no confiables, depende de la compatibilidad de versiones de Python y librerias, no ofrece trazabilidad ni versionado por si mismo y no es el formato mas portable para servir modelos en distintos entornos.
4. Si quisieramos servir varias versiones a la vez, separariamos la resolucion del modelo de la logica del endpoint. Por ejemplo, cargando un registro `version -> modelo`, exponiendo la version en la URL o en un parametro y dejando cada modelo serializado en una ruta distinta con su propia metadata.
