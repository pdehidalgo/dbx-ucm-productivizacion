# Ejercicio práctico: Airflow Training Pipeline

En este ejercicio vas a construir una version reducida del **DAG de entrenamiento** de `airflow_project` para que el flujo funcione end-to-end con **Docker + Airflow + MLflow**. El foco no esta en refactorizar el proyecto completo, sino en completar un ejemplo didáctico y verificar que deja trazabilidad real en MLflow.

La copia del ejercicio usa un `dag_id` especifico, `ml_pipeline_preprocessing_training_exercise`, para no colisionar con el DAG real del proyecto.

## Objetivos

1. Identificar la configuración mínima del pipeline de entrenamiento.
2. Completar `ReaderFactory` y `TrainerFactory` para soportar el flujo base.
3. Entrenar un modelo de clasificación y registrar métricas y artifacts en MLflow.
4. Implementar el registro condicional del modelo comparando contra el histórico.
5. Ejecutar el DAG desde Airflow y validar el run resultante en MLflow.

````python
from pathlib import Path
import json
import textwrap


def find_project_root() -> Path:
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / "airflow_project").exists() and (candidate / "notebooks_clase").exists():
            return candidate
    raise FileNotFoundError("No se encontro la raiz del proyecto")


PROJECT_ROOT = find_project_root()
AIRFLOW_PROJECT = PROJECT_ROOT / "airflow_project"
DAG_PATH = AIRFLOW_PROJECT / "dags" / "airflow_ml_pipeline_completar.py"
CONFIG_PATH = AIRFLOW_PROJECT / "config" / "config.yaml"
VARIABLES_PATH = AIRFLOW_PROJECT / "config" / "variables.json"

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DAG del ejercicio:", DAG_PATH)
print("Config:", CONFIG_PATH)
print("Variables:", VARIABLES_PATH)
````
